```python
#List(ordered,mutable,allows duplication)
student_list=["Amith","banu","priya","bharath"]
print(student_list)
```

```python
student_list=["Amith","banu","priya","bharath"]
student_list.append("adi")
print(student_list)
```

```python
student_list=["Amith","banu","priya","bharath"]
student_list.insert(3,"pratham")
print(student_list)
```

```python
student_list=["Amith","banu","priya","bharath"]
student_list.extend([101,102])
print(student_list)
```

```python
student_list=["Amith","banu","priya","bharath",102]
student_list.remove(102)
print(student_list)
```

```python
student_list=["Amith","banu","priya","bharath",102,103,104]
student_list.pop()
print(student_list)
```

```python
student_list=["Amith","banu","priya","bharath",102,103,104]
print(len(student_list))
```

```python
student_list=["Amith","banu","priya","bharath",102,103,104]
student_list.reverse()
print(student_list)
```

```python
var=[10,20,50,30,20,20,20,20,30,40,50,60,70,80,90]
print(var.count(20))
print(max(var))
print(min(var))
print(sum(var))
print(var.index(90))
```

```python
#tuple(ordered,immutable,allows duplication)
var=(10,20,30,40,50,60,70,80,90)
print(var)
```

```python
var=(10,20,30,40,50,60,70,80,90)
print(var)
print(var.count(20))
print(max(var))
print(min(var))
print(sum(var))
print(var.index(90))
```

```python
#set(unordered,mutable,no duplicate elements)
var={10,20,30,40,50,60,70,80,90}
print(var)
```

```python
var={10,20,30,40,50,60,70,80,90}
print(var[1])
#because set is unordered
```

```python
var={10,20,30,40,50,60,70,80,90,20,50,20,30}
print(var)
#removes duplicate
```

```python
var={"python","java","node"}
var.add("HTML")
print(var)
```

```python
var={"python","java","node"}
var.add("node")#does not allow duplicate
print(var)
```

```python
var={"python","java","node"}
var.update("html","c","c++")
print(var)
```

```python
var={"python","java","node","html","c","c++"}
var.remove("html")
print(var)
```

```python
var={"python","java","node","c","c++"}
var.remove("html")
print(var)
```

```python
var={"python","java","node","c","c++"}
var.discard("html")
print(var)
```

```python
var={"python","java","node","c","c++"}
var.pop()
print(var)
```

```python
#union
var1={10,20,30,40,50,60}
var2={60,70,80,90,100}
var_name=var1.union(var2)
print(var_name)
```

```python
#intersection
var1={10,20,30,40,50,60}
var2={60,70,80,90,100}
var_name=var1.intersection(var2)
print(var_name)
```
