```python
class BankAccount:
    bank_name = "SBI"
    total_accounts = 0
    def __init__(self,holder, balance):
        self.holder = holder
        self.balance = balance
        BankAccount.total_accounts += 1
    def show(self):
        print(f"Account holder is {self.holder} and balance is {self.balance} and holder account is at {BankAccount.bank_name}")

account1 = BankAccount("Rohan", 100000)
account2 = BankAccount("ROhit", 200000)

account1.show()
account2.show()

print("Total accounts oepned ", BankAccount.total_accounts)
```

```python
# inheritance
class Vehicle:
    def start(self):
        print("Vehicle starting ")

class Car(Vehicle):
    def drive(self):
        print("Car driving")

c1 = Car()
c1.start()
c1.drive()

```

```python
#multilevel

class Person:
    def __init__(self, name):
        self.name = name
class Employee(Person):
    def __init__(self,name, emp_id):
        super().__init__(name)
        self.emp_id = emp_id

class Manager(Employee):
    def __init__(self, name, emp_id, team_size):
        super().__init__(name, emp_id)
        self.team_size = team_size

    def details(self):
        print(f"{self.name} (ID: {self.emp_id}) manages a team of {self.team_size}")

m = Manager("Priya",101, 25)
m.details()

```

```python
# multiple

class DeliveryVehicle:
    def vehicle_info(self):
        print("uses of two wheeler for delivery")

class ZomatoPartnerBase:
    def rating_info(self):
        print("Maintain atleast 4 rating")

class DeliveryPartner(DeliveryVehicle, ZomatoPartnerBase):
    def __init__(self, name):
        self.name = name

obj = DeliveryPartner("Rohan")
obj.vehicle_info()
obj.rating_info()
```

```python
# Encapsulations
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def salary_detils(self):
        return self.salary

emp = Employee("AMit", 90000)
print(emp.salary_detils())
print(emp.salary)
emp.salary = 1200000
print(emp.salary)
```

```python
class ATM:
    def __init__(self, pin, balance):
        self.__pin = pin
        self.__balance = balance

    def withdraw(self,entered_pin, amount):
        if entered_pin != self.__pin:
            print("Wrong Pin")
            return
        if amount > self.__balance:
            print("Insufficient fund")
            return
        self.__balance -= amount
        print(f"Withdraw {amount} sucessfully, Remaing balance is {self.__balance}")

a = ATM(1234, 50000)
a.withdraw(8798, 40000)
a.withdraw(1234, 70000)
a.withdraw(1234, 40000)
```

```python
# Polymorphism
#method overloading
class Programming_language:
    def usage(self):
        print("General Usage of programming langauge")

class Python_lang(Programming_language):
    def usage(self):
        print("Python is used for the AI and web application")
class Java(Programming_language):
    def usage(self):
        print("Java is used for desktop and web application")
class C_lang(Programming_language):
    def usage(self):
        print("C is used for desktop application")


obj = Python_lang()
obj.usage()
obj = C_lang()
obj.usage()
```

```python
class Cricket_score:
    def __init__(self, run):
        self.run = run
    def __add__(self, other):
        return Cricket_score(self.run - other.run)

    def __str__(self):
        return f"{self.run} runs"


inning1 = Cricket_score(120)
inning2 = Cricket_score(140)
total = inning1 + inning2
print("Match total score is : ", total)
```
