```python
#LOGICAL OPERATOR
#and
#or
#not
```

```python
#and
age=25
print(age<18 and age<60)
print(age<60 and age<18)
print(age>18 and age<60)
```

```python
#or
age=25
print(age<18 or age<60)
print(age<60 or age<18)
print(age>18 or age<60)
```

```python
#not
age=25
print(not(age < 18 or age > 60))
print(not(age < 60 or age < 18))
print(not(age > 18 or age < 60))
```
