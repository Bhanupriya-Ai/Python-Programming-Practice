```python
#simulation of atm pin
attempted=0
correct_pin=8765
entered_pin=0000
while entered_pin!=correct_pin and attempted<3:
  entered_pin=int(input("enter pin"))
  attempted=attempted+1
  if entered_pin==correct_pin:
    print("pin verified")
  else:
    print("wrong pin")
    if attempted==3 and entered_pin!=correct_pin:
      print("ur card is blocked")
```
