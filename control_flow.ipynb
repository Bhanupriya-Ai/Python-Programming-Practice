```python
#nested if
#railway ticket booking
age=int(input("enter age"))
adhar_no=int(input("enter last four digit of adhar if you have otherwise 0"))
if age>=18:
  if adhar_no!=0:
    print("Ticket booking is allowed")
  else:
    print("ticket booking is not allowed because u dont hv adhar")
else:
  print("ticket booking is not allowed because ur age is less than 18")
```
